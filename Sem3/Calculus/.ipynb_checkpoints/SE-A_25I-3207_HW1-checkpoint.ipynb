```python
# Import NumPy for creating numerical input values and calculations.
import numpy as np

# Import pyplot for drawing and displaying graphs.
import matplotlib.pyplot as plt

# Create 1000 x-values from -5 to 15 for the memory function.
x1 = np.linspace(-5, 15, 1000)

# Calculate the expression inside the square root.
inside = 100 + 20*x1 - x1**2

# Keep only the x-values for which the square root is real.
valid = inside >= 0

# Compute the memory function at all valid x-values.
y1 = np.sqrt(inside[valid])

# Create x-values on both sides of x = 4 for the rational function.
x2_left = np.linspace(-10, 3.9, 500)
x2_right = np.linspace(4.1, 15, 500)

# Compute the rational function on the left of x = 4.
y2_left = (3*x2_left + 2)/(x2_left - 4)

# Compute the rational function on the right of x = 4.
y2_right = (3*x2_right + 2)/(x2_right - 4)

# Create input values from 0 to 4 for the compression-error function.
x3 = np.linspace(0, 4, 500)

# Compute the absolute-value compression-error function.
y3 = np.abs(x3**2 - 4*x3 + 3)

# Create one figure containing three graphs arranged in one row.
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))

# Draw the memory-usage graph in the first position.
axes[0].plot(x1[valid], y1, color="blue", linewidth=2)

# Add a title to the first graph.
axes[0].set_title(r"$M(x)=\sqrt{100+20x-x^2}$")

# Label the horizontal axis of the first graph.
axes[0].set_xlabel("x")

# Label the vertical axis of the first graph.
axes[0].set_ylabel("M(x)")

# Draw horizontal and vertical coordinate axes on the first graph.
axes[0].axhline(0, color="black", linewidth=0.8)
axes[0].axvline(0, color="black", linewidth=0.8)

# Add a grid to the first graph.
axes[0].grid(True, alpha=0.3)

# Draw both branches of the rational function in the second position.
axes[1].plot(x2_left, y2_left, color="red", linewidth=2)
axes[1].plot(x2_right, y2_right, color="red", linewidth=2)

# Show the excluded input x = 4 with a dashed vertical line.
axes[1].axvline(4, color="black", linestyle="--", linewidth=1)

# Add the title and axis labels to the second graph.
axes[1].set_title(r"$T(x)=\frac{3x+2}{x-4}$")
axes[1].set_xlabel("x")
axes[1].set_ylabel("T(x)")

# Restrict the displayed y-values so the graph is easy to read.
axes[1].set_ylim(-20, 20)

# Draw the coordinate axes and grid on the second graph.
axes[1].axhline(0, color="black", linewidth=0.8)
axes[1].grid(True, alpha=0.3)

# Draw the compression-error graph in the third position.
axes[2].plot(x3, y3, color="purple", linewidth=2)

# Add the title and axis labels to the third graph.
axes[2].set_title(r"$E(x)=|x^2-4x+3|$")
axes[2].set_xlabel("x")
axes[2].set_ylabel("E(x)")

# Draw the coordinate axes and grid on the third graph.
axes[2].axhline(0, color="black", linewidth=0.8)
axes[2].axvline(0, color="black", linewidth=0.8)
axes[2].grid(True, alpha=0.3)

# Adjust the spacing so titles and labels do not overlap.
plt.tight_layout()

# Display all three graphs.
plt.show()
```

```python
# Import NumPy for numerical values.
import numpy as np

# Import pyplot for graphing.
import matplotlib.pyplot as plt

# Create 100 study-time values from 0 to 5 hours.
x = np.linspace(0, 5, 100)

# Use the regression equation to predict the quiz scores.
y = 4*x + 3

# Plot the predicted score against study time.
plt.plot(x, y, color="darkgreen", linewidth=2)

# Add a descriptive title to the graph.
plt.title("Regression Model: Predicted Quiz Score")

# Label the horizontal and vertical axes.
plt.xlabel("Study time (hours)")
plt.ylabel("Predicted quiz score")

# Add a grid to make values easier to read.
plt.grid(True, alpha=0.3)

# Display the completed graph.
plt.show()
```
